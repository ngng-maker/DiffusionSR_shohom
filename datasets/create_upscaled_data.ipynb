```python
# from PIL import Image
import requests
import matplotlib.pyplot as plt
import numpy as np
from scipy import interpolate
import os
import glob
import cv2
from skimage.transform import resize
import pandas
import h5py
from tqdm import tqdm
```

```python
def resize_datapoint(lr_img, method = 'cubic'):
    upscaled = cv2.resize(lr_img, (factor*20,factor*20), interpolation=method)
    return upscaled
def test_resize(lr_img, up_img):
    if len(lr_img.shape)  == 3:
        for i in range(lr_img.shape[2]):
            plt.imshow(lr_img[:,:, i].T, origin = 'lower')
            plt.title("LR")
            plt.show()
            plt.title("HR")
            plt.imshow(up_img[:,:, i].T, origin = 'lower')
            plt.show()
    else:
        plt.imshow(lr_img.T, origin = 'lower')
        plt.title("LR")
        plt.show()
        plt.title("HR")
        plt.imshow(up_img.T, origin = 'lower')
        plt.show()
factor = 2
root_folder = '/home/oogoke/DiffusionSR/datasets/expanded_ss316l_all_laser_velocity_xz_cross_section_data'#'/home/oogoke/DiffusionSR/datasets/simulation_basis_v3_laser_velocity_xz_cross_section_data'#'/home/oogoke/DiffusionSR/datasets/simulation_basis_laser_velocity_xz_cross_section_data'
method_folders = glob.glob(os.path.join(root_folder, "**/LR/*"))
for folder in method_folders:
    os.makedirs(os.path.join(folder, '{}x').format(factor), exist_ok = True)
```

```python
datapoints = glob.glob(os.path.join(root_folder, "**/LR/*/*/*/*npy"))
for i in tqdm(range(len(datapoints))):
    original = np.load(datapoints[i])
    upscaled = resize_datapoint(original, cv2.INTER_CUBIC)
    original_fname = datapoints[i]
    newfname = datapoints[i].replace('1x', '{}x'.format(factor))

    new_dir = '/'.join(newfname.split('/')[:-1])
    if not os.path.isdir(new_dir):
        os.mkdir(new_dir)
    np.save(newfname, upscaled)
    # break
    # if i % 1000 == 0:
        # print(newfname)
```
