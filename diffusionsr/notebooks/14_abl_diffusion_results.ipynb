````python
%matplotlib inline
import os, sys, time
from pathlib import Path
import numpy as np
import torch
from torch.utils.data import DataLoader
from scipy.ndimage import gaussian_filter as _gf, binary_erosion
from scipy.spatial import KDTree
import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display as _ipy_display

_plt_show_orig = plt.show
def _show(*a, **kw):
    for n in plt.get_fignums(): _ipy_display(plt.figure(n))
    plt.close('all')
plt.show = _show

if not torch.cuda.is_available() and DEVICE == 'cuda':
    print('No GPU — falling back to CPU'); DEVICE = 'cpu'

def find_root(start=Path.cwd()):
    for p in [start, *start.parents]:
        if (p/'setup.py').exists() and (p/'diffusionsr').exists(): return p
    raise RuntimeError('Could not find project root')
PROJECT_ROOT = find_root()
if str(PROJECT_ROOT) not in sys.path: sys.path.insert(0, str(PROJECT_ROOT))
print(f'PROJECT_ROOT: {PROJECT_ROOT} | device: {DEVICE}')
````

````python
from diffusionsr.datasets.dataset import SimulationXZDataset
from diffusionsr.analysis.analysis_functions import get_profile, load_encoder

def as_numpy(x): return x.detach().cpu().numpy() if isinstance(x, torch.Tensor) else np.asarray(x)
def mae_rmse(p, g):
    p, g = np.asarray(p).ravel(), np.asarray(g).ravel()
    return {'MAE': float(np.mean(np.abs(p-g))), 'RMSE': float(np.sqrt(np.mean((p-g)**2)))}

def boundary_pixels(mask): return np.argwhere(mask & ~binary_erosion(mask, structure=np.ones((3,3))))
def chamfer_dist(a, b):
    ba, bb = boundary_pixels(a), boundary_pixels(b)
    if len(ba)==0 or len(bb)==0: return float('nan')
    return float((KDTree(bb).query(ba)[0].mean() + KDTree(ba).query(bb)[0].mean()) / 2)

def _consistency_metrics(bin_t, bin_liq):
    """IOU / MSE / Chamfer between two binary masks (T>T_LIQ vs liquid field)."""
    inter = (bin_t & bin_liq).sum()
    union = (bin_t | bin_liq).sum()
    iou = inter / union if union > 0 else float('nan')
    mse = float(np.mean((bin_t.astype(float) - bin_liq.astype(float)) ** 2))
    return iou, mse, chamfer_dist(bin_t, bin_liq)

_PLATE_IDX = round(PLATE_HEIGHT_UM / VOXEL_SIZE_UM[1])
_VOXEL_A   = VOXEL_SIZE_UM[0] * VOXEL_SIZE_UM[1]

def _physical_mp_metrics(temp2d, sdf2d=None):
    """depth_um, area_um2, keyhole_depth_um from 2D arrays (length × depth)."""
    result = {}
    mask = temp2d[:, :_PLATE_IDX] > MELT_THRESHOLD
    if mask.any():
        min_z = int(np.argwhere(mask)[:, 1].min())
        result['depth_um'] = float(PLATE_HEIGHT_UM - (min_z + 0.5) * VOXEL_SIZE_UM[1])
        result['area_um2'] = float(mask.sum() * _VOXEL_A)
    else:
        result['depth_um'], result['area_um2'] = 0.0, 0.0
    if sdf2d is not None:
        kh = sdf2d[:, :_PLATE_IDX] < 0
        if kh.any():
            kh_z = int(np.argwhere(kh)[:, 1].min())
            result['keyhole_depth_um'] = float(PLATE_HEIGHT_UM - (kh_z + 0.5) * VOXEL_SIZE_UM[1])
        else:
            result['keyhole_depth_um'] = 0.0
    return result

def load_datasets(v):
    kw = dict(downscale_method=DOWNSCALE_METHOD, root_folder=DATA_ROOT,
              normalize=v['normalize'], n_steps=N_STEPS, field_names=v['field_names'])
    train_ds, dev_ds, test_ds = (SimulationXZDataset(split=s, **kw) for s in ['train','dev','test'])
    return train_ds, dev_ds, test_ds

def _liq_mask(phys, fn, ch_offset=0):
    if 'sdfliqlabel' in fn: return phys[ch_offset + fn.index('sdfliqlabel')] < 0
    if 'liqlabel'    in fn: return phys[ch_offset + fn.index('liqlabel')]    > LIQ_THR
    return np.zeros(phys.shape[1:], dtype=bool)

def _mk_contour(ax, phys, fn, sigma=1.5, ch_offset=0):
    if 'temperature' not in fn: return
    if not ('sdfliqlabel' in fn or 'liqlabel' in fn): return
    try:
        ax.contour(_gf(_liq_mask(phys, fn, ch_offset=ch_offset).T.astype(float), sigma), levels=[0.5],
                   colors=['white'], linewidths=[1.], origin='lower', alpha=0.85)
    except Exception: pass

def plot_comparison(panels, title, fn, ch=0):
    all_vals = np.concatenate([np.asarray(d).ravel() for _, d, _ in panels])
    vmin, vmax = float(np.nanpercentile(all_vals, 2)), float(np.nanpercentile(all_vals, 98))
    fig, axes = plt.subplots(1, len(panels), figsize=(4.5*len(panels), 4), dpi=120)
    if len(panels) == 1: axes = [axes]
    for ax, (ttl, data, phys) in zip(axes, panels):
        ax.imshow(np.asarray(data).T, origin='lower', cmap='jet', vmin=vmin, vmax=vmax, aspect='auto')
        _mk_contour(ax, phys, fn, ch_offset=ch)
        ax.set_title(ttl, fontsize=9); ax.axis('off')
    plt.suptitle(title, fontsize=10); plt.tight_layout(); plt.show()

def plot_loss_curves(v, model_label, enc_project='RRDN_Encoder', model_project='Flow3D_SuperResolution'):
    import wandb
    api = wandb.Api()
    entity = os.environ.get('WANDB_ENTITY', 'ngng-')
    fig, ax = plt.subplots(figsize=(9, 4), dpi=120); plotted = False
    if v.get('enc_wb_name'):
        runs = list(api.runs(f"{entity}/{enc_project}", filters={"display_name": v['enc_wb_name']}))
        if runs:
            for col, ls in [("train_loss","-"),("test_loss","--")]:
                h = runs[0].history(keys=[col], samples=10000, pandas=True)
                if col in h.columns:
                    ax.plot(h[col].dropna().values, ls=ls, label=f'Enc:{col}', lw=1.5, alpha=0.9); plotted=True
    if v.get('wandb_run_name'):
        runs = list(api.runs(f"{entity}/{model_project}", filters={"display_name": v['wandb_run_name']}))
        if runs:
            for col, ls in [("train_loss","-"),("val_loss","--")]:
                h = runs[0].history(keys=[col], samples=10000, pandas=True)
                if col in h.columns:
                    ax.plot(h[col].dropna().values, ls=ls, label=col, lw=1.5, alpha=0.9); plotted=True
    if plotted:
        ax.set_xlabel('Epoch'); ax.set_ylabel('Loss')
        ax.set_title(f"{v['run_name']} — Training Curves"); ax.legend(fontsize=8); ax.grid(True, alpha=0.3)
        plt.tight_layout(); plt.show()
    else:
        plt.close(fig)
        print(f"No W&B data found for '{v['run_name']}'. Run may still be training or wandb_run_name may not match.")

def load_wandb_preds(v):
    import wandb
    api = wandb.Api()
    entity = os.getenv('WANDB_ENTITY', 'ngng-')
    artifact = api.artifact(f"{entity}/Flow3D_SuperResolution/{v['run_name']}_eval_predictions:latest")
    d = np.load(os.path.join(artifact.download(), 'predictions.npz'), allow_pickle=True)
    return d['pred'], d['gt']

def summarise_metrics(v, preds, gts, fn, ch=0, model_label='Model'):
    maes, rmses, mp_errs, kh_errs = [], [], [], []
    depth_p, depth_g = [], []
    area_p,  area_g  = [], []
    khdep_p, khdep_g = [], []
    iou_p,   iou_g   = [], []
    mse_p,   mse_g   = [], []
    cham_p,  cham_g  = [], []
    has_liq = 'sdfliqlabel' in fn or 'liqlabel' in fn
    liq_fn  = 'sdfliqlabel' if 'sdfliqlabel' in fn else ('liqlabel' if 'liqlabel' in fn else None)
    last_base = (N_STEPS - 1) * len(fn)
    t_ch = last_base + fn.index('temperature') if 'temperature' in fn else ch
    l_ch = last_base + fn.index(liq_fn) if liq_fn else None

    for p, g in zip(preds, gts):
        p_ch, g_ch = p[ch], g[ch]
        m = mae_rmse(p_ch, g_ch); maes.append(m['MAE']); rmses.append(m['RMSE'])
        try:
            pmp, pkh = get_profile(p[ch:ch+1]); gmp, gkh = get_profile(g[ch:ch+1])
            mp_errs.append(float(np.mean(np.abs(pmp-gmp)))); kh_errs.append(float(np.mean(np.abs(pkh-gkh))))
        except: pass
        try:
            p_sdf = p[l_ch] if (has_liq and liq_fn == 'sdfliqlabel') else None
            g_sdf = g[l_ch] if (has_liq and liq_fn == 'sdfliqlabel') else None
            pm = _physical_mp_metrics(p[t_ch], sdf2d=p_sdf)
            gm = _physical_mp_metrics(g[t_ch], sdf2d=g_sdf)
            depth_p.append(pm['depth_um']); depth_g.append(gm['depth_um'])
            area_p.append(pm['area_um2']);  area_g.append(gm['area_um2'])
            if 'keyhole_depth_um' in pm:
                khdep_p.append(pm['keyhole_depth_um']); khdep_g.append(gm['keyhole_depth_um'])
        except: pass
        if has_liq and l_ch is not None:
            try:
                bin_tp = p[t_ch] > T_LIQ
                bin_lp = (p[l_ch] < 0) if liq_fn == 'sdfliqlabel' else (p[l_ch] > LIQ_THR)
                ip, mp, cp = _consistency_metrics(bin_tp, bin_lp)
                iou_p.append(ip); mse_p.append(mp); cham_p.append(cp)
                bin_tg = g[t_ch] > T_LIQ
                bin_lg = (g[l_ch] < 0) if liq_fn == 'sdfliqlabel' else (g[l_ch] > LIQ_THR)
                ig, mg, cg = _consistency_metrics(bin_tg, bin_lg)
                iou_g.append(ig); mse_g.append(mg); cham_g.append(cg)
            except: pass

    print(f"[{v['run_name']}] n={len(maes)}")
    print(f"  MAE  = {np.nanmean(maes):.4f} ± {np.nanstd(maes):.4f}")
    print(f"  RMSE = {np.nanmean(rmses):.4f} ± {np.nanstd(rmses):.4f}")
    if mp_errs:  print(f"  MP-MAE       = {np.nanmean(mp_errs):.2f} ± {np.nanstd(mp_errs):.2f} px")
    if kh_errs:  print(f"  KH-MAE       = {np.nanmean(kh_errs):.2f} ± {np.nanstd(kh_errs):.2f} px")
    if depth_p:  print(f"  Depth MAE    = {np.nanmean(np.abs(np.array(depth_p)-np.array(depth_g))):.2f} μm")
    if area_p:   print(f"  Area MAE     = {np.nanmean(np.abs(np.array(area_p)-np.array(area_g))):.2f} μm²")
    if khdep_p:  print(f"  KH-Depth MAE = {np.nanmean(np.abs(np.array(khdep_p)-np.array(khdep_g))):.2f} μm")
    if iou_p:    print(f"  IOU  pred={np.nanmean(iou_p):.4f}  gt={np.nanmean(iou_g):.4f}")
    if mse_p:    print(f"  MSE  pred={np.nanmean(mse_p):.6f}  gt={np.nanmean(mse_g):.6f}")
    if cham_p:   print(f"  Chamfer pred={np.nanmean(cham_p):.3f} px  gt={np.nanmean(cham_g):.3f} px")

    def _nm(lst): return float(np.nanmean(lst)) if lst else float('nan')
    def _mae2(a, b): return float(np.nanmean(np.abs(np.array(a)-np.array(b)))) if a else float('nan')
    return dict(
        run_name=v['run_name'],
        mae=_nm(maes), rmse=_nm(rmses),
        mp_mae=_nm(mp_errs), kh_mae=_nm(kh_errs),
        depth_mae=_mae2(depth_p, depth_g),
        area_mae=_mae2(area_p, area_g),
        kh_depth_mae=_mae2(khdep_p, khdep_g),
        iou_pred=_nm(iou_p), mse_pred=_nm(mse_p), chamfer_pred=_nm(cham_p),
        iou_gt=_nm(iou_g),   mse_gt=_nm(mse_g),   chamfer_gt=_nm(cham_g),
    )
````

````python
from diffusionsr.runners.train_diffusion import DiffusionModel

def load_diffusion(v):
    train_ds, dev_ds, test_ds = load_datasets(v)
    enc = load_encoder(v['enc_dir'], train_ds) if v['encoding'] else None
    model = DiffusionModel(
        results_folder=v['run_dir'], lr_encoder_folder=v.get('enc_dir'),
        train_dataset=train_ds, dev_dataset=dev_ds, test_dataset=test_ds,
        timesteps=TIMESTEPS, conditioning=v['conditioning'], encoding=v['encoding'],
        schedule=SCHEDULE, device=DEVICE,
    )
    model.load_saved_model()
    fn = v['field_names']
    print(f"Loaded {v['run_name']}  fields={fn}  normalize={v['normalize']}")
    return train_ds, dev_ds, test_ds, enc, model
````

## 3. DDPM vs DDIM Sampler Comparison


````python
# ── DDIM skip ablation — loads from W&B artifacts (no GPU/data needed) ──
# Run slurm/ablation_20260901/06_step_abl_array.sh on TRACE to generate artifacts.
import wandb as _wandb

SKIP_VALUES = [5, 10, 20, 50, 100]   # + 'ddpm' for full DDPM
V_ABL = VARIANTS['diff_enc_multifield_std']  # change to probe a different variant
fn_abl = V_ABL['field_names']
VIZ_CH_ABL = (N_STEPS - 1) * len(fn_abl)

_api = _wandb.Api()
_entity = os.getenv('WANDB_ENTITY', 'ngng-')

abl_results = {}   # skip_val (int or 'ddpm') -> (mae, mp_mae)
_missing = []
for _sk in SKIP_VALUES + ['ddpm']:
    _aname = f"{V_ABL['run_name']}_ddim_abl_{_sk}"
    try:
        _art = _api.artifact(f"{_entity}/Flow3D_SuperResolution/{_aname}:latest")
        _d = np.load(os.path.join(_art.download(), 'predictions.npz'), allow_pickle=True)
        _preds, _gts = _d['pred'], _d['gt']
        _maes, _mps = [], []
        for _p, _g in zip(_preds, _gts):
            _maes.append(mae_rmse(_p[VIZ_CH_ABL], _g[VIZ_CH_ABL])['MAE'])
            try:
                _pmp, _ = get_profile(_p[VIZ_CH_ABL:VIZ_CH_ABL+1])
                _gmp, _ = get_profile(_g[VIZ_CH_ABL:VIZ_CH_ABL+1])
                _mps.append(float(np.mean(np.abs(_pmp - _gmp))))
            except: pass
        abl_results[_sk] = (float(np.nanmean(_maes)), float(np.nanmean(_mps)) if _mps else float('nan'))
        _label = f"skip={_sk}" if isinstance(_sk, int) else 'DDPM'
        print(f"  {_label:12s}: MAE={abl_results[_sk][0]:.4f}  MP-MAE={abl_results[_sk][1]:.2f} px  (n={len(_maes)})")
    except Exception as _e:
        _missing.append(_sk)
        print(f"  skip={_sk}: artifact not found — {_e}")

if _missing:
    print(f"\nMissing skip values {_missing}: run slurm/ablation_20260901/06_step_abl_array.sh on TRACE.")

if len(abl_results) >= 2:
    def _to_steps(s): return TIMESTEPS // s if isinstance(s, int) else TIMESTEPS
    _ordered = sorted(abl_results, key=_to_steps, reverse=True)
    _steps_v = [_to_steps(s) for s in _ordered]
    _mae_v   = [abl_results[s][0] for s in _ordered]
    _mp_v    = [abl_results[s][1] for s in _ordered]

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5), dpi=120)
    ax1.plot(_steps_v, _mae_v, 'o-', color='royalblue', lw=2, ms=8)
    ax1.set_xlabel('Denoising steps (1000 = DDPM)'); ax1.set_ylabel('MAE')
    ax1.set_title('DDIM Ablation — Field MAE'); ax1.grid(True, alpha=0.3)
    ax2.plot(_steps_v, _mp_v, 's-', color='green', lw=2, ms=8)
    ax2.set_xlabel('Denoising steps'); ax2.set_ylabel('MP-MAE (px)')
    ax2.set_title('DDIM Ablation — MP Depth'); ax2.grid(True, alpha=0.3)
    plt.suptitle(f"{V_ABL['run_name']} — DDIM step sensitivity", fontsize=11)
    plt.tight_layout(); plt.show()
elif not abl_results:
    print(f"No DDIM skip ablation artifacts found for {V_ABL['run_name']}.")
    print("Run slurm/ablation_20260901/06_step_abl_array.sh (tasks 16-23) on TRACE.")
````
